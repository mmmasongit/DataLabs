**SELECT** is the most common statement used, and it allows us to retrieve\
information from a table.

Example syntax for **SELECT** statement:

```sql
SELECT column_name
FROM table_name;
```

Execution order is **FROM table_1** and then **SELECT c1**

Not good practice to use \* to **SELECT** all columns unless you need all\
columns. It will query everything and increase traffic between server and the\
application, which can result in slowing down of retrieval of results.


## Select all columns


````python
%%sql
SELECT *
FROM actor;
````

## Select by columns


````python
%%sql
SELECT
  first_name, last_name
FROM
  actor;
````
