# Notes

Sometimes a table contains a column that has duplicate values, and you may\
want to only select list that is unique/distinct.

The **DISTINCT** keyword can be used to return only the distinct values in a\
column.

The **DISTINCT** keyword operates on a column. The syntax looks like this:\

```sql
SELECT
  DISTINCT column_name
FROM
  table_name;
```

Notice example 1 how `Adam` is repeated twice. We can get rid of all the\
entries that are repeated using **DISTINCT**. By using DISTINCT we will only\
get 1 adam.


## Example 1


````python
%%sql
SELECT 
  first_name
FROM 
  actor
WHERE 
  first_name = 'Adam';
````

````python
%%sql
SELECT
  DISTINCT first_name
FROM
  actor
WHERE
  first_name = 'Adam';
````
